## Topic 4: Conditional Statements

### Question 1

```python
# Answer 1
# Write your answer here

temp = 25

if temp == 20:
    print("Warm")
else:
    print("Cool")
```

## Topic 5: Loops

### Question 1
Use a `for` loop to print each item in `[10, 20, 30]`.

```python
# Answer 1
# Write your answer here

list = [10, 20, 30];

for i in list:
    print(i)
```

### Question 2
Use a loop to calculate the sum of `[2, 4, 6]`.

```python
# Answer 2
# Write your answer here

list = [2, 4, 6]
total = 0

for i in list:
    total += i

print(total)
```

### Question 3
Count how many values in `[5, 12, 8, 20]` are greater than 10.

```python
# Answer 3
# Write your answer here

list = [5, 12, 8, 20]
count = 0

for i in list:
    if i > 10:
        count += 1
print(count)
```
