```python

# If/else statement 

amount = 108
if amount > 100:
    print("Amount is greater then 100")
else:
    print("Amount is less then 100")
```

```python

# If/elif/else statement 
amount = 108
if amount > 100:
    print("Amount is greater then 100")
elif amount == 108:
    print("Amount is 108")
else:
    print("Amount is less then 100")

# when we use elif statement conditions must be written based on priority
```

```python
# for loop
 
prices = [11,32,52,108,111]
total = 0

for i in prices:
    print("i is", i)
    print(f"Im adding {i} in {total}")
    total = total + i
    print(f"total till {i} is {total}")
print(f"final total is", total)
```

```python
# reusing same thing again and again

# without fun
a = 2
b = 3
c = a + b
print(c)

# With function
""""Syntax
def function_name(parameter1, parameter2, parameter3):
    # function body
    return result
"""

def add(a,b):
    print("Adding", a, "and", b)
    c = a+b
    print("Result is", c)

add(11,32)


def add(a,b):
    return a + b
print(add(2,3))
print(add(10,11))
print(add(100,11))


# function with default value
def add(a,b=1):
    return a + b
print(add(2))
print(add(2,3))
print(add(10))
print(add(10,11))
```
